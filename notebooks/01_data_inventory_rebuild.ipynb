```python
import sqlite3
import zipfile
import pandas as pd
from pathlib import Path

RAW_DIR = PROJECT_DIR / "data" / "raw"

# Temporary audit database
DB_PATH = PROJECT_DIR / "data" / "processed" / "ride_id_audit.sqlite"

conn = sqlite3.connect(DB_PATH)

conn.execute("""
CREATE TABLE IF NOT EXISTS ride_ids (
    ride_id TEXT PRIMARY KEY,
    first_archive TEXT
)
""")

conn.execute("DELETE FROM ride_ids")
conn.commit()

duplicate_summary = []

for archive in sorted(RAW_DIR.glob("*.zip")):

    archive_duplicates = 0

    with zipfile.ZipFile(archive, "r") as z:

        csv_name = next(
            name for name in z.namelist()
            if name.lower().endswith(".csv")
            and not name.startswith("__MACOSX/")
            and not Path(name).name.startswith("._")
        )

        with z.open(csv_name) as file:

            for chunk in pd.read_csv(
                file,
                usecols=["ride_id"],
                chunksize=100_000,
                dtype={"ride_id": "string"}
            ):

                ids = chunk["ride_id"].dropna()

                # Count repeated IDs within this chunk
                chunk_duplicates = ids.duplicated().sum()

                # Count IDs already encountered in previous chunks/files
                existing = conn.execute(
                    "SELECT ride_id FROM ride_ids WHERE ride_id IN ({})".format(
                        ",".join("?" for _ in ids.unique())
                    ),
                    ids.unique().tolist()
                ).fetchall() if len(ids.unique()) else []

                existing_ids = {row[0] for row in existing}

                # Count all duplicate occurrences
                previously_seen = ids.isin(existing_ids).sum()

                archive_duplicates += (
                    int(chunk_duplicates) + int(previously_seen)
                )

                # Insert only previously unseen IDs
                new_ids = ids[
                    ~ids.isin(existing_ids)
                    & ~ids.duplicated()
                ]

                conn.executemany(
                    """
                    INSERT OR IGNORE INTO ride_ids
                    (ride_id, first_archive)
                    VALUES (?, ?)
                    """,
                    [
                        (ride_id, archive.name)
                        for ride_id in new_ids
                    ]
                )

                conn.commit()

    duplicate_summary.append({
        "archive": archive.name,
        "duplicate_occurrences": archive_duplicates
    })

    print(
        archive.name,
        "— duplicate occurrences:",
        archive_duplicates
    )

duplicate_df = pd.DataFrame(duplicate_summary)

print("\nTotal duplicate occurrences:",
      duplicate_df["duplicate_occurrences"].sum())

display(duplicate_df)
```

### Duplicate Ride ID Investigation

The investigation identified 35 distinct `ride_id` values appearing in both the April 2026 and May 2026 archives.

Each ID appeared twice, resulting in 70 matching records across the two archives.

A full-record comparison was performed using all 13 original dataset columns. All 35 pairs were identical, with no differences detected.

**Findings:**
- 35 distinct duplicated ride IDs.
- 70 records involved in the comparison.
- 35 identical pairs across the April and May archives.
- No conflicting records identified.

**Proposed cleaning decision:**

Retain one record per `ride_id` when constructing the combined analytical dataset. Exclude the additional copy to prevent double-counting.

The original ZIP archives will remain unchanged.

**Limitation:**

The investigation confirms that the records are identical but does not establish why the data provider included them in both monthly archives.

```python
import sqlite3
import zipfile
import pandas as pd

# Connect to the existing audit database
DB_PATH = PROJECT_DIR / "data" / "processed" / "ride_id_audit.sqlite"
conn = sqlite3.connect(DB_PATH)

# Count occurrences of each ride_id in the May archive
conn.execute("""
CREATE TEMP TABLE IF NOT EXISTS may_counts (
    ride_id TEXT PRIMARY KEY,
    occurrences INTEGER
)
""")
conn.execute("DELETE FROM may_counts")

may_archive = RAW_DIR / "202605-divvy-tripdata.zip"

with zipfile.ZipFile(may_archive) as zf:
    csv_name = next(
        name for name in zf.namelist()
        if name.endswith(".csv")
        and "__MACOSX" not in name
        and not name.split("/")[-1].startswith("._")
    )

    with zf.open(csv_name) as f:
        for chunk in pd.read_csv(
            f,
            usecols=["ride_id"],
            dtype={"ride_id": "string"},
            chunksize=100_000
        ):
            counts = chunk["ride_id"].dropna().value_counts()

            conn.executemany(
                """
                INSERT INTO may_counts (ride_id, occurrences)
                VALUES (?, ?)
                ON CONFLICT(ride_id)
                DO UPDATE SET occurrences =
                    occurrences + excluded.occurrences
                """,
                [(str(ride_id), int(count))
                 for ride_id, count in counts.items()]
            )

conn.commit()

# Identify IDs already seen in earlier archives
# or repeated within the May archive
duplicate_ids = pd.read_sql_query("""
SELECT m.ride_id, m.occurrences, r.first_archive
FROM may_counts m
JOIN ride_ids r ON m.ride_id = r.ride_id
WHERE m.occurrences > 1
   OR r.first_archive != '202605-divvy-tripdata.zip'
ORDER BY m.ride_id
""", conn)

print("Distinct duplicate ride IDs:", len(duplicate_ids))
display(duplicate_ids)
```

```python

# MISSING-VALUE AUDIT ACROSS ALL 12 ARCHIVES

missing_summary = {}
total_rows = 0

for archive in sorted(RAW_DIR.glob("*.zip")):

    with zipfile.ZipFile(archive) as zf:

        csv_name = next(
            name for name in zf.namelist()
            if name.endswith(".csv")
            and "__MACOSX" not in name
            and not name.split("/")[-1].startswith("._")
        )

        with zf.open(csv_name) as f:

            for chunk in pd.read_csv(
                f,
                chunksize=100_000
            ):

                total_rows += len(chunk)

                for column, count in chunk.isna().sum().items():
                    missing_summary[column] = (
                        missing_summary.get(column, 0) + count
                    )


# Create the missing-value summary
missing_df = pd.DataFrame({
    "column": list(missing_summary.keys()),
    "missing_values": list(missing_summary.values())
})

missing_df["missing_percentage"] = (
    missing_df["missing_values"] / total_rows * 100
).round(4)

missing_df = missing_df.sort_values(
    "missing_values",
    ascending=False
).reset_index(drop=True)

print(f"Total records inspected: {total_rows:,}")
print("\nMissing-value summary:")

display(missing_df)
```

```python

# PAIRED MISSING-VALUE INVESTIGATION

columns_to_check = [
    "start_station_name",
    "start_station_id",
    "end_station_name",
    "end_station_id",
    "end_lat",
    "end_lng"
]

pair_counts = {
    "start_station": {
        "both_missing": 0,
        "name_only_missing": 0,
        "id_only_missing": 0,
        "neither_missing": 0
    },
    "end_station": {
        "both_missing": 0,
        "name_only_missing": 0,
        "id_only_missing": 0,
        "neither_missing": 0
    },
    "end_coordinates": {
        "both_missing": 0,
        "latitude_only_missing": 0,
        "longitude_only_missing": 0,
        "neither_missing": 0
    }
}

for archive in sorted(RAW_DIR.glob("*.zip")):

    with zipfile.ZipFile(archive) as zf:

        csv_name = next(
            name for name in zf.namelist()
            if name.endswith(".csv")
            and "__MACOSX" not in name
            and not name.split("/")[-1].startswith("._")
        )

        with zf.open(csv_name) as f:

            for chunk in pd.read_csv(
                f,
                usecols=columns_to_check,
                chunksize=100_000
            ):

                # Starting station pair
                start_name = chunk["start_station_name"].isna()
                start_id = chunk["start_station_id"].isna()

                pair_counts["start_station"]["both_missing"] += (
                    (start_name & start_id).sum()
                )

                pair_counts["start_station"]["name_only_missing"] += (
                    (start_name & ~start_id).sum()
                )

                pair_counts["start_station"]["id_only_missing"] += (
                    (~start_name & start_id).sum()
                )

                pair_counts["start_station"]["neither_missing"] += (
                    (~start_name & ~start_id).sum()
                )

                # Ending station pair
                end_name = chunk["end_station_name"].isna()
                end_id = chunk["end_station_id"].isna()

                pair_counts["end_station"]["both_missing"] += (
                    (end_name & end_id).sum()
                )

                pair_counts["end_station"]["name_only_missing"] += (
                    (end_name & ~end_id).sum()
                )

                pair_counts["end_station"]["id_only_missing"] += (
                    (~end_name & end_id).sum()
                )

                pair_counts["end_station"]["neither_missing"] += (
                    (~end_name & ~end_id).sum()
                )

                # Ending coordinate pair
                end_lat = chunk["end_lat"].isna()
                end_lng = chunk["end_lng"].isna()

                pair_counts["end_coordinates"]["both_missing"] += (
                    (end_lat & end_lng).sum()
                )

                pair_counts["end_coordinates"]["latitude_only_missing"] += (
                    (end_lat & ~end_lng).sum()
                )

                pair_counts["end_coordinates"]["longitude_only_missing"] += (
                    (~end_lat & end_lng).sum()
                )

                pair_counts["end_coordinates"]["neither_missing"] += (
                    (~end_lat & ~end_lng).sum()
                )


# Display the results
paired_missing_df = pd.DataFrame([
    {
        "field_group": group,
        "missing_pattern": pattern,
        "record_count": count
    }
    for group, patterns in pair_counts.items()
    for pattern, count in patterns.items()
])

display(paired_missing_df)
```

### Paired Missing-Value Investigation

An additional investigation was conducted to determine whether missing values in related dataset columns occurred independently or together.

**Findings:**

| Field group | Both missing | Neither missing |
|---|---:|---:|
| Starting station | 1,289,486 | 4,882,299 |
| Ending station | 1,355,487 | 4,816,298 |
| Ending coordinates | 5,451 | 6,166,334 |

No records were found with only one missing value within any of the three field pairs.

Specifically:
- `start_station_name` and `start_station_id` were always missing together.
- `end_station_name` and `end_station_id` were always missing together.
- `end_lat` and `end_lng` were always missing together.

**Analytical implications:**

Missing station information limits station-level analysis but does not invalidate the affected trips for broader analyses of rider behaviour, ride duration, time patterns or bike preferences.

**Proposed handling:**
- Preserve records with missing station information.
- Do not invent or impute station names, IDs or coordinates.
- Assess station-level analyses using records with valid station information.
- Retain trips with missing ending coordinates for analyses that do not require those coordinates.

No records have been modified or deleted during this investigation.

```python

# MISSING-VALUE DISTRIBUTION BY MONTH, RIDER CATEGORY
# AND BIKE TYPE

grouped_results = []

for archive in sorted(RAW_DIR.glob("*.zip")):

    with zipfile.ZipFile(archive) as zf:

        csv_name = next(
            name for name in zf.namelist()
            if name.endswith(".csv")
            and "__MACOSX" not in name
            and not name.split("/")[-1].startswith("._")
        )

        with zf.open(csv_name) as f:

            for chunk in pd.read_csv(
                f,
                usecols=[
                    "ride_id",
                    "started_at",
                    "member_casual",
                    "rideable_type",
                    "start_station_name",
                    "end_station_name",
                    "end_lat"
                ],
                chunksize=100_000
            ):

                # Assign month using the actual trip start date
                chunk["start_month"] = (
                    pd.to_datetime(
                        chunk["started_at"],
                        errors="coerce"
                    )
                    .dt.strftime("%Y-%m")
                )

                # Create missing-value indicators
                chunk["start_station_missing"] = (
                    chunk["start_station_name"].isna()
                )

                chunk["end_station_missing"] = (
                    chunk["end_station_name"].isna()
                )

                chunk["end_coordinates_missing"] = (
                    chunk["end_lat"].isna()
                )

                # Group by month, rider category and bike type
                grouped = (
                    chunk.groupby(
                        [
                            "start_month",
                            "member_casual",
                            "rideable_type"
                        ],
                        dropna=False
                    )
                    .agg(
                        total_records=("ride_id", "size"),
                        start_station_missing=(
                            "start_station_missing", "sum"
                        ),
                        end_station_missing=(
                            "end_station_missing", "sum"
                        ),
                        end_coordinates_missing=(
                            "end_coordinates_missing", "sum"
                        )
                    )
                    .reset_index()
                )

                grouped_results.append(grouped)


# Combine the partial summaries
missing_distribution = (
    pd.concat(grouped_results, ignore_index=True)
    .groupby(
        ["start_month", "member_casual", "rideable_type"],
        dropna=False
    )
    .sum(numeric_only=True)
    .reset_index()
)

# Calculate missing-value percentages
for column in [
    "start_station",
    "end_station",
    "end_coordinates"
]:
    missing_distribution[f"{column}_missing_pct"] = (
        missing_distribution[f"{column}_missing"]
        / missing_distribution["total_records"]
        * 100
    ).round(2)

missing_distribution = missing_distribution.sort_values(
    ["start_month", "member_casual", "rideable_type"]
).reset_index(drop=True)

display(missing_distribution)
```

The duration audit identified 29 records with negative calculated trip durations. All 29 records originated from the November 2025 archive and had both timestamps recorded within the 1:00 a.m. hour on 2 November 2025.

Further investigation identified 391 trips with both start and end timestamps within this repeated hour:

- 362 records had positive calculated durations.
- No records had zero calculated duration.

This pattern is consistent with the end of daylight-saving time in Chicago on 2 November 2025, when clocks moved backward from 2:00 a.m. to 1:00 a.m.

Because the dataset timestamps do not distinguish between the two occurrences of the repeated hour, the actual elapsed duration cannot be reliably reconstructed from these timestamp values alone.

**Treatment decision:** Preserve all 391 records in the source data. Flag this cohort as potentially affected by daylight-saving-time ambiguity and exclude it from duration-based analysis unless reliable timezone information becomes available. Do not automatically adjust the 29 negative durations.

### Long-Duration Trip Investigation — Trips Exceeding 24 Hours

The duration audit identified 5,324 trips exceeding 24 hours. All records involved classic bikes.

Further investigation revealed a strong concentration around 25 hours:

- 5,312 trips lasted between 24 and 25 hours.
- 12 trips exceeded 25 hours.
- 5,203 trips (97.73%) were within one minute of 25 hours.

An examination of station information showed that 5,216 records (97.97%) had missing end-station IDs, while only 108 had complete start- and end-station IDs.

The median duration was approximately 24.998 hours for records with missing end-station information, compared with 24.434 hours for records with available end-station information. This association warrants further investigation but does not establish the operational reason behind the pattern.

**Daylight-saving-time investigation**

Ten of the 12 trips exceeding 25 hours started on 7 March 2026 and ended on 8 or 9 March 2026. This period coincided with Chicago's daylight-saving-time transition on 8 March 2026, when clocks moved forward by one hour.

Because the dataset does not provide timezone offsets, the actual elapsed durations of these records cannot be reliably reconstructed from the timestamps alone.

The remaining two trips occurred in September 2026 and lasted approximately 25 hours and one minute each. No daylight-saving explanation has been established for these records.

**Treatment decision:** Preserve all 5,324 records in the master dataset. Flag the long-duration cohort for investigation. Separately flag the ten March records as potentially affected by daylight-saving-time ambiguity. Exclude the ten ambiguous records from primary duration-based analysis provisionally, unless reliable timezone information becomes available.

Do not automatically correct or delete any of these records. Retain them for analyses where their duration values do not compromise the result.

### Verification Against Official Divvy Documentation

The official Divvy system-data documentation confirms that published trip data excludes staff trips associated with servicing and inspecting the system. It also states that trips shorter than 60 seconds are removed because they may represent false starts or attempts to re-dock a bike.

However, the published documentation reviewed does not specify a maximum trip duration of 24 or 25 hours, nor does it explain the concentration of records around 25 hours.

Consequently, the 5,324 trips exceeding 24 hours cannot be classified as invalid based on the available documentation alone.

**Decision:** Retain these records in the master dataset and flag them for investigation. Do not attribute the near-25-hour pattern to a particular operational or technical cause without further evidence.

```python

import zipfile
import sqlite3
import pandas as pd

# --------------------------------------------------
# 1. SAFETY CHECKS
# --------------------------------------------------

if MASTER_FILE.exists():
    raise FileExistsError(
        f"Master dataset already exists: {MASTER_FILE}"
    )

TEMP_MASTER_FILE = PROCESSED_DIR / "cyclistic_master.csv.tmp"

if TEMP_MASTER_FILE.exists():
    raise FileExistsError(
        f"Temporary output already exists: {TEMP_MASTER_FILE}. "
        "Inspect it before proceeding."
    )

# --------------------------------------------------
# 2. CONNECT TO THE AUDIT DATABASE
# --------------------------------------------------

connection = sqlite3.connect(db_path)

# Expected unique ride IDs per archive
archive_expected = pd.read_sql_query(
    """
    SELECT first_archive, COUNT(*) AS expected_count
    FROM ride_ids
    GROUP BY first_archive
    """,
    connection
)

expected_counts = dict(
    zip(
        archive_expected["first_archive"],
        archive_expected["expected_count"]
    )
)

# --------------------------------------------------
# 3. INITIALIZE CONSOLIDATION TRACKING
# --------------------------------------------------

total_retained = 0
file_written = False
archive_results = []

# --------------------------------------------------
# 4. PROCESS EACH ARCHIVE
# --------------------------------------------------

try:

    for archive in archive_files:

        archive_name = archive.name

        print(f"\nProcessing: {archive_name}")

        # Retrieve IDs assigned to this archive
        archive_ids = set(
            pd.read_sql_query(
                """
                SELECT ride_id
                FROM ride_ids
                WHERE first_archive = ?
                """,
                connection,
                params=(archive_name,)
            )["ride_id"]
        )

        expected_count = expected_counts[archive_name]
        archive_retained = 0

        with zipfile.ZipFile(archive, "r") as zip_file:

            csv_files = [
                name for name in zip_file.namelist()
                if name.lower().endswith(".csv")
                and not name.startswith("__MACOSX/")
                and not name.split("/")[-1].startswith("._")
            ]

            if len(csv_files) != 1:
                raise ValueError(
                    f"Expected one main CSV in {archive_name}; "
                    f"found {len(csv_files)}."
                )

            with zip_file.open(csv_files[0]) as csv_stream:

                chunk_reader = pd.read_csv(
                    csv_stream,
                    chunksize=CHUNK_SIZE,
                    low_memory=False
                )

                for chunk_number, chunk in enumerate(
                    chunk_reader, start=1
                ):

                    if chunk.columns.tolist() != column_names:
                        raise ValueError(
                            f"Column mismatch in {archive_name}."
                        )

                    # Retain only IDs assigned to this archive
                    retained_chunk = chunk[
                        chunk["ride_id"].isin(archive_ids)
                    ]

                    if not retained_chunk.empty:

                        retained_chunk.to_csv(
                            TEMP_MASTER_FILE,
                            mode="a" if file_written else "w",
                            header=not file_written,
                            index=False
                        )

                        file_written = True

                    archive_retained += len(retained_chunk)

        # --------------------------------------------------
        # 5. VERIFY THIS ARCHIVE
        # --------------------------------------------------

        if archive_retained != expected_count:
            raise ValueError(
                f"Count mismatch for {archive_name}: "
                f"expected {expected_count:,}, "
                f"retained {archive_retained:,}."
            )

        total_retained += archive_retained

        archive_results.append({
            "archive": archive_name,
            "expected_records": expected_count,
            "retained_records": archive_retained,
            "difference": archive_retained - expected_count
        })

        print(
            f"Verified: {archive_retained:,} records retained."
        )

    # --------------------------------------------------
    # 6. VERIFY TOTAL BEFORE FINALIZING
    # --------------------------------------------------

    if total_retained != unique_id_count:
        raise ValueError(
            f"Final count mismatch: expected {unique_id_count:,}, "
            f"retained {total_retained:,}."
        )

    if not file_written:
        raise ValueError("No records were written.")

    # Rename temporary output to final master dataset
    TEMP_MASTER_FILE.replace(MASTER_FILE)

    print("\nCONSOLIDATION COMPLETED")
    print("Final dataset:", MASTER_FILE)
    print(f"Total records: {total_retained:,}")

except Exception:
    print(
        "\nConsolidation stopped. "
        "The final master filename has not been created."
    )
    raise

finally:
    connection.close()

# --------------------------------------------------
# 7. DISPLAY ARCHIVE-BY-ARCHIVE RESULTS
# --------------------------------------------------

archive_results_df = pd.DataFrame(archive_results)

display(archive_results_df)

print("\nTotal records retained:", f"{total_retained:,}")
print("Expected total:", f"{unique_id_count:,}")
```

```python

import pandas as pd

# --------------------------------------------------
# 1. DEFINE THE ANALYTICAL DATASET PATH
# --------------------------------------------------

ANALYTICAL_FILE = PROCESSED_DIR / "cyclistic_master.parquet"

# Prevent accidental overwriting
if ANALYTICAL_FILE.exists():
    raise FileExistsError(
        f"Analytical dataset already exists: {ANALYTICAL_FILE}"
    )

print("Source CSV:", MASTER_FILE)
print("Destination Parquet:", ANALYTICAL_FILE)

# --------------------------------------------------
# 2. CONVERT CSV TO PARQUET IN CHUNKS
# --------------------------------------------------

parquet_writer = None
converted_rows = 0

try:

    for chunk_number, chunk in enumerate(
        pd.read_csv(
            MASTER_FILE,
            chunksize=CHUNK_SIZE,
            low_memory=False
        ),
        start=1
    ):

        # Convert timestamp columns to datetime
        chunk["started_at"] = pd.to_datetime(
            chunk["started_at"]
        )

        chunk["ended_at"] = pd.to_datetime(
            chunk["ended_at"]
        )

        # Convert each chunk to a PyArrow table
        import pyarrow as pa
        import pyarrow.parquet as pq

        arrow_table = pa.Table.from_pandas(
            chunk,
            preserve_index=False
        )

        # Create the writer from the first chunk
        if parquet_writer is None:

            parquet_writer = pq.ParquetWriter(
                ANALYTICAL_FILE,
                arrow_table.schema,
                compression="snappy"
            )

        # Append each chunk
        parquet_writer.write_table(arrow_table)

        converted_rows += len(chunk)

        print(
            f"Chunk {chunk_number}: "
            f"{converted_rows:,} rows converted"
        )

finally:

    if parquet_writer is not None:
        parquet_writer.close()

# --------------------------------------------------
# 3. REPORT CONVERSION RESULTS
# --------------------------------------------------

print("\nCONVERSION COMPLETED")
print(f"Rows converted: {converted_rows:,}")

print(
    "Parquet file size:",
    f"{ANALYTICAL_FILE.stat().st_size / (1024**2):,.2f} MB"
)
```

```python
import pyarrow.parquet as pq
import pandas as pd

# Open the Parquet file
parquet_file = pq.ParquetFile(ANALYTICAL_FILE)

# Expected structure from the original dataset
expected_columns = [
    "ride_id",
    "rideable_type",
    "started_at",
    "ended_at",
    "start_station_name",
    "end_station_name",
    "start_station_id",
    "end_station_id",
    "start_lat",
    "start_lng",
    "end_lat",
    "end_lng",
    "member_casual"
]

# Expected values from our validated master CSV
expected_rows = 6_171_750

expected_missing = {
    "start_station_name": 1_289_480,
    "start_station_id": 1_289_480,
    "end_station_name": 1_355_473,
    "end_station_id": 1_355_473,
    "end_lat": 5_445,
    "end_lng": 5_445
}

# Initialize validation counters
actual_rows = parquet_file.metadata.num_rows
actual_columns = parquet_file.schema_arrow.names

missing_counts = {column: 0 for column in expected_columns}
rider_counts = {}
bike_counts = {}

# Read the Parquet file in batches to avoid loading everything into memory
for batch in parquet_file.iter_batches(batch_size=100_000):

    # Count missing values in every column
    for column in expected_columns:
        missing_counts[column] += batch.column(
            column
        ).null_count

    # Count rider categories
    batch_df = batch.select(
        ["member_casual", "rideable_type"]
    ).to_pandas()

    rider_values = batch_df["member_casual"].value_counts()
    bike_values = batch_df["rideable_type"].value_counts()

    for category, count in rider_values.items():
        rider_counts[category] = rider_counts.get(category, 0) + count

    for category, count in bike_values.items():
        bike_counts[category] = bike_counts.get(category, 0) + count


# Display validation results
print("PARQUET VALIDATION RESULTS")
print("=" * 50)

print(f"Expected rows: {expected_rows:,}")
print(f"Actual rows:   {actual_rows:,}")
print(f"Row count matches: {actual_rows == expected_rows}")

print(f"\nExpected columns: {len(expected_columns)}")
print(f"Actual columns:   {len(actual_columns)}")
print(f"Column names match: {actual_columns == expected_columns}")

print("\nPARQUET DATA TYPES")
print(parquet_file.schema_arrow)

print("\nMISSING VALUE VALIDATION")
for column, expected in expected_missing.items():
    actual = missing_counts[column]
    print(
        f"{column}: {actual:,} "
        f"| Expected: {expected:,} "
        f"| Match: {actual == expected}"
    )

print("\nRIDER CATEGORY COUNTS")
print(rider_counts)

print("\nBIKE CATEGORY COUNTS")
print(bike_counts)

print("\nVALIDATION COMPLETE")
```
